```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('../data/raw/bedtime_screentime_sleep_debt.csv')
```

```python
df.head()
```

```python
df.info()
```

```python
plt.figure(figsize=(5,4)) # Размер графика
plt.title('Распредление возраста') # Заголовок
ax = sns.histplot(
    df,
    x='age',
    binwidth=5,
)
plt.show()
```

```python
plt.figure(figsize=(5,4)) # Размер графика
plt.title('Распредление возраста') # Заголовок
ax = sns.histplot(
    df,
    x='age',
    binwidth=4,
    hue='gender',
)
plt.show()
```

```python
plt.figure(figsize=(5,2)) # Размер графика
plt.title('Ящик с усами для возраста') # Заголовок
ax = sns.boxplot(
    df,
    x='age',
)
plt.show()
```

```python
plt.figure(figsize=(10,4)) # Размер графика
plt.title('Ящик с усами для Продолжительность сна') # Заголовок
ax = sns.boxplot(
    df,
    x='total_sleep_hours',
    hue='gender',
)
sns.move_legend(ax, "center right")
plt.show()
```

```python
plt.figure(figsize=(10,4)) # Размер графика
plt.title('Диаграмма рассеивания для Продолжительность сна и Время засыпания') # Заголовок
ax = sns.scatterplot(
    df,
    x='sleep_latency_min',
    y='total_sleep_hours',
    # hue='gender',
)
plt.show()
```

```python
plt.figure(figsize=(10,4)) # Размер графика
plt.title('Столбчатая диаграмма для Время засыпания и Категории дефецита сна') # Заголовок
ax = sns.barplot(
    df,
    x='sleep_debt_category',
    y='sleep_latency_min',
    estimator='mean',
    hue='blue_light_filter_active'
)
plt.show()
```

```python
plt.figure(figsize=(10,4)) # Размер графика
plt.title('Столбчатая диаграмма для Время засыпания и Категории дефецита сна') # Заголовок
ax = sns.barplot(
    df,
    y='sleep_debt_category',
    x='sleep_latency_min',
    estimator='median',
    hue='blue_light_filter_active'
)
plt.show()
```
