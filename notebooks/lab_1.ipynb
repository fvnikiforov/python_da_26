# Задание 1 (тестовые задачи)
```
1. is_even(n)
Вернуть True, если n четное

2. last_char(s)
Вернуть последний символ строки s. Если пустой текст, вернуть пустой текст

3. sum_of_squares(xs)
Сумма квадратов всех чисел из списка xs

4. unique_preserve_order(xs)
Удалить дубликаты из текстового списка xs, сохранив порядок

5. word_count(text)
Верните частоты слов: разберите по пробелам, приведите к нижнему регистру, уберите простую пунктуацию .,;:!? на концах слова

6. reverse_words(s)
Разверните порядок слов в строке s
```
