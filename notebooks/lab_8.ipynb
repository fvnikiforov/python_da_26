```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv(
    '../data/raw/insurance.csv',
    dtype = {
        "sex":"category",
        "smoker":"category",
        "region":"category",
    }
)
```

```python
# bmi < 18.5 then under weight
# bmi >= 18.5 and bmi <= 25.0 then normal weight
# bmi > 25.0 then excess weight
def classify_bmi(bmi:float)->str:
    if bmi < 18.5:
        return 'under weight'
    elif bmi >= 18.5 and bmi <= 25.0:
        return 'normal weight'
    elif bmi > 25.0:
        return 'excess weight'
    else:
        return np.nan
    
```

```python
df['bmi_category'] = df['bmi'].apply(classify_bmi)
```

```python
# bmi < 18.5 then under weight
# bmi >= 18.5 and bmi <= 25.0 then normal weight
# bmi > 25.0 then excess weight
conditions = [
    df['bmi'] < 18.5,
    df['bmi'] <= 25.0,
    df['bmi'] > 25.0,
]

choices = [
    'under weight',
    'normal weight',
    'excess weight',
]
df['bmi_category_2'] = np.select(conditions, choices, default=None)
```

```python
df[['bmi', 'bmi_category', 'bmi_category_2']]
```

```python
# smoker = 'no' and bmi_category = 'normal' then 'cheep charges'
# else 'expenses charges'
def classify_charges(row):
    if (row['smoker'] == 'no') and (row['bmi_category'] == 'normal weight'):
        return 'cheep charges'
    else:
        return 'expenses charges'
```

```python
df['charges_category'] = df.apply(classify_charges, axis=1)
```

```python
df
```

```python
df['charges_category_2'] = np.where(
    (df['smoker'] == 'no') & (df['bmi_category'] == 'normal weight'),
    'cheep charges',
    'expenses charges',
)
```
