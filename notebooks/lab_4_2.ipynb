```python
import pandas as pd
```

```python
df_pop = pd.read_csv('../data/raw/gapminder/pop.csv')
```

```python
df_murder = pd.read_csv('../data/raw/gapminder/murder_total_deaths.csv')
```

```python
df_pop
```

```python
df_pop.info()
```

```python
df_pop = pd.melt(
    df_pop,
    id_vars=['geo', 'name'],
    var_name='year',
    value_name='population',
)
```

```python
df_pop
```

```python
df_pop.info()
```

```python
df_pop = df_pop.dropna()
df_pop = df_pop.reset_index(drop=True)
```

```python
df_pop['year'] = pd.to_numeric(df_pop['year'])
```

```python
df_pop.info()
```

```python
df_murder = pd.melt(
    df_murder,
    id_vars=['geo', 'name'],
    var_name='year',
    value_name='murder_total_deaths',
)
```

```python
df_murder = df_murder.dropna()
df_murder = df_murder.reset_index(drop=True)
```

```python
df_murder['year'] = pd.to_numeric(df_murder['year'])
```

```python
df_murder.info()
```

```python
df_murder
```

```python
df_merged = pd.merge(
    left=df_pop,
    right=df_murder,
    how='inner', # inner, left, right, outer, cross
    # left_on=['geo', 'name', 'year'],
    # right_on=['geo', 'name', 'year'],
    on=['geo', 'name', 'year'],
)
```

```python
df_merged
```

```python
df_merged.info()
```

```python
df_merged[df_merged['geo']=='rus']
```

```python
df_merged['murder_in_100k_pop'] = df_merged['murder_total_deaths'] / df_merged['population'] * 100000
```

```python
df_merged
```

```python
pd.pivot_table(
    data=df_merged,
    index=['geo', 'name'],
    values=['population', 'murder_in_100k_pop'],
    aggfunc=['mean', 'min', 'max'],
).sort_values(('mean', 'murder_in_100k_pop'))
```
