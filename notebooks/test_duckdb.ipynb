```python
%%sql
SELECT
    object_name
    , education_level
    , specialty
    , year
    , gender
    , average_salary_fact_avg
    , average_salary_fact_med
FROM 
    read_parquet('../data/raw/data_study_area_125_v20250709_csv/data_graduates_specialty_125_v20250709.parquet')
where 
    object_name = 'Республика Саха (Якутия)'
    and specialty = 'Прикладная информатика'
    and education_level = 'Бакалавриат, специалитет'
order by
    year,
    gender
;
```
