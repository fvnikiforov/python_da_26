```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('../data/raw/mcdonalds_menu_20260827_202242.csv')
```

```python
df.head()
```

```python
df.info()
```

```python
df['serving_size'].unique()
```

```python
plt.figure(figsize=(5,4)) # Размер графика
plt.title('Распредление Калорий') # Заголовок
ax = sns.histplot(
    df,
    x='calories',
    binwidth=100,
)
plt.show()
```

```python
plt.figure(figsize=(5,4)) # Размер графика
plt.title('Распредление Калорий') # Заголовок
ax = sns.histplot(
    df,
    x='calories',
    binwidth=100,
    hue='is_popular',
)
plt.show()
```

```python
plt.figure(figsize=(5,2)) # Размер графика
plt.title('Ящик с усами для Калорий') # Заголовок
ax = sns.boxplot(
    df,
    x='calories',
)
plt.show()
```

```python
plt.figure(figsize=(5,4)) # Размер графика
plt.title('Ящик с усами для Калорий') # Заголовок
ax = sns.boxplot(
    df,
    x='calories',
    y='category',
)
plt.show()
```

```python
plt.figure(figsize=(10,4)) # Размер графика
plt.title('Диаграмма рассеивания для Калорий и Стоимости') # Заголовок
ax = sns.scatterplot(
    df,
    x='calories',
    y='price_usd',
)
plt.show()
```

```python
plt.figure(figsize=(10,4)) # Размер графика
plt.title('Диаграмма рассеивания для Калорий и Стоимости') # Заголовок
ax = sns.scatterplot(
    df,
    x='calories',
    y='price_usd',
    hue='is_popular',
)
plt.show()
```

```python
plt.figure(figsize=(10,4)) # Размер графика
plt.title('Столбчатая диаграмма для Категории блюда и Стоимости') # Заголовок
ax = sns.barplot(
    df,
    x='category',
    y='price_usd',
    estimator='mean',
)
plt.show()
```

```python
plt.figure(figsize=(10,4)) # Размер графика
plt.title('Столбчатая диаграмма для Категории блюда и Стоимости') # Заголовок
ax = sns.barplot(
    df,
    x='category',
    y='price_usd',
    estimator='mean',
    hue='is_popular'
)
plt.show()
```
