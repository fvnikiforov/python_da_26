```python
import pandas as pd
```

```python
df_pop = pd.read_csv('../data/raw/gapminder/pop.csv')
```

```python
df_ppl_poverty = pd.read_csv('../data/raw/gapminder/number_of_people_in_poverty.csv')
```

```python
df_pop
```

```python
df_pop = pd.melt(
    df_pop,
    id_vars=['geo', 'name'],
    var_name='year',
    value_name='population',
)
```

```python
df_pop
```

```python
df_pop.info()
```

```python
df_pop['year'] = pd.to_numeric(df_pop['year'])
```

```python
df_pop.info()
```

```python
df_pop[(df_pop['name']=='Russia') & (df_pop['year']==2020)]
```

```python
df_ppl_poverty = pd.melt(
    df_ppl_poverty,
    id_vars=['geo', 'name'],
    var_name='year',
    value_name='people_in_poverty',
)
```

```python
df_ppl_poverty
```

```python
df_ppl_poverty.info()
```

```python
df_ppl_poverty = df_ppl_poverty.dropna(ignore_index=True)
```

```python
df_ppl_poverty['year'] = pd.to_numeric(df_ppl_poverty['year'])
```

```python
df_ppl_poverty.info()
```

```python
df_merged = pd.merge(
    left=df_pop,
    right=df_ppl_poverty,
    how='inner',
    left_on=['geo', 'name', 'year'],
    right_on=['geo', 'name', 'year'],
)
```

```python
df_merged
```

```python
df_merged.info()
```

```python
df_merged['share_poverty'] = df_merged['people_in_poverty'] / df_merged['population'] * 100
```

```python
df_merged[df_merged['share_poverty']<0]
```

```python
pd.pivot_table(
    df_merged[df_merged['year']>=2000],
    index=['geo', 'name'],
    values=['people_in_poverty', 'share_poverty'],
    aggfunc=['mean', 'min', 'max'],
).sort_values(('mean', 'share_poverty'))
```
