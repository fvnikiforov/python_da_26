```python
translate_dict = {
    'voronezh':'воронеж',
    'vladivostok':'владивосток',
    'ufa':'уфа',
    'chita':'чита',
    'ulyanovsk':'ульяновск',
    'novosibirsk':'новосибирск',
}
```

```python
df['city_clean'] = df['city_low']
```

```python
for eng, rus in translate_dict.items():
    df['city_clean'] = df['city_clean'].str.replace(eng, rus)
```
