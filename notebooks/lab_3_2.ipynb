```python
import pandas as pd
```

```python
df = pd.read_csv('../data/raw/tax_revenue_percent_of_gdp.csv')
```

```python
df.head()
```

```python
df_melt = pd.melt(
    df,
    id_vars=['geo', 'name'],
    var_name='year',
    value_name='tax_revenue_percent_of_gdp'
)
```

```python
df_melt
```

```python
df_melt.info()
```

```python
df_melt.notna()
```

```python
df_melt.isna()
```

```python
df_melt.isna().sum()
```

```python
df_melt['tax_revenue_percent_of_gdp'] = df_melt['tax_revenue_percent_of_gdp'].fillna(0)
```

```python
df_melt = df_melt.fillna(0)
```

```python
df_clean = df_melt.dropna(
    # axis=0,  # 0, 1
    # how='any',  # any, all
    # thresh=4,
    # subset='tax_revenue_percent_of_gdp',
    # subset=['geo', 'name', 'tax_revenue_percent_of_gdp'],
    # ignore_index=True
)
```

```python
df_clean = df_clean.reset_index(drop=True)
```

```python
df_clean
```

```python
df_clean.info()
```

```python
pd.to_numeric(
    df_clean['year'],
    downcast='integer'
)
```

```python
df_clean['year'] = pd.to_numeric(df_clean['year'])
```

```python
df_clean
```

```python
df_clean.info()
```

```python
df_clean.groupby(by=['geo', 'name']).agg('mean')
```

```python
df_clean[['name', 'tax_revenue_percent_of_gdp']].groupby(by='name').agg('mean')
```

```python
df_clean[['name', 'tax_revenue_percent_of_gdp']].groupby(by='name').agg('mean').sort_values(by='tax_revenue_percent_of_gdp')
```

```python
df_grp = df_clean[['name', 'tax_revenue_percent_of_gdp']].groupby(by='name').agg(['mean', 'median', 'min', 'max', 'count'])
```

```python
df_grp
```
